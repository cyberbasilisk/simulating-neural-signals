```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
# Initialize Parameters
sampling_rate = 1000
time_vector = np.arange(-1, 1, 1/sampling_rate)
timepoints = len(time_vector)
```

```python
# Generate A Cosine Wave And A Gaussian

freqeuncy = 20
fwhm = 0.1

cosine_wave = np.cos(2 * np.pi * freqeuncy * time_vector )
gaus = np.exp(-4 * np.log(2) * (time_vector ** 2) / (fwhm ** 2))
```

```python
# Visualize Cosine Wave And Gaussian

plt.close('all')
plt.subplots(1, 1, figsize=(10, 4))
plt.plot(time_vector, cosine_wave, c='g')
plt.plot(time_vector, gaus, c='r')
plt.xlabel('Time')
plt.ylabel('Amplitude')
plt.legend(['Cosine', 'Gaussian'])
plt.show()
```

```python
# Taper Gaussian To Cosine Wave To Create A Morlet Wavelet
m_wavlet = cosine_wave * gaus
```

```python
# Visualize Wavelet
plt.close('all')
plt.subplots(1, 1, figsize=(10, 4))
plt.plot(time_vector, m_wavlet, c='black')
plt.xlabel('Time')
plt.ylabel('Amplitude')
plt.show()
```
