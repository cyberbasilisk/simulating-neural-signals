```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
# Initialize Dimensions

trials = 30
channels = 16
sampling_rate = 500
timepoints = sampling_rate * 2
time_vector = np.arange(timepoints) / sampling_rate
```

```python
# Initialize Chirps Variable

chirps = np.zeros((channels, timepoints, trials))
chirps.shape
```

```python
# Generating Chirps With A Frequency Modulator Of 5 Hz - 50 Hz Range

frequency_modulator = np.linspace(5, 50, timepoints)

for i in range(channels):
  for k in range(trials):
    chirps[i, :, k] = np.sin(2 * np.pi * (time_vector + (np.cumsum(frequency_modulator) / sampling_rate)))
```

```python
# Visualizing Chirps

plt.plot(time_vector, chirps[0, :, 0])
plt.xlabel('Time (s)')
plt.ylabel('Amplitude (uV)')
plt.title('Data')
plt.show()
```
